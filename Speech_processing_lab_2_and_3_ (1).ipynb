```python
# 1. Load and preprocess audio (mono + resample to 16kHz)
import librosa
import numpy as np
import torch
import matplotlib.pyplot as plt
import soundfile as sf

from transformers import Wav2Vec2Processor, Wav2Vec2ForCTC

audio_path = "/content/drive/MyDrive/Speech processing/LJ001-0001.wav"

# Load audio
audio, sr = librosa.load(audio_path, sr=None, mono=True)

# Resample to 16kHz if needed
target_sr = 16000
if sr != target_sr:
    audio = librosa.resample(audio, orig_sr=sr, target_sr=target_sr)
    sr = target_sr

# 2. Load pre-trained Wav2Vec2 model
processor = Wav2Vec2Processor.from_pretrained("facebook/wav2vec2-base-960h")
model = Wav2Vec2ForCTC.from_pretrained("facebook/wav2vec2-base-960h")

# 3. Phoneme recognition (CTC-based)
inputs = processor(audio, sampling_rate=sr, return_tensors="pt", padding=True)

with torch.no_grad():
    logits = model(inputs.input_values).logits

predicted_ids = torch.argmax(logits, dim=-1)
tokens = processor.tokenizer.convert_ids_to_tokens(predicted_ids[0])

# 4. Estimate phoneme time intervals
time_per_step = audio.shape[0] / logits.shape[1] / sr

phonemes = []
start = 0
prev = tokens[0]

for i, tok in enumerate(tokens):
    if tok != prev:
        end = i
        if prev != "|":  # ignore silence token
            phonemes.append((prev, start * time_per_step, end * time_per_step))
        start = i
        prev = tok

# Add last phoneme
end = len(tokens)
if prev != "|":
    phonemes.append((prev, start * time_per_step, end * time_per_step))

# 5–6. Extract, save, and visualize phoneme segments
output_segments = []

for idx, (ph, t_start, t_end) in enumerate(phonemes[:5]):  # first 5 phonemes
    s = int(t_start * sr)
    e = int(t_end * sr)
    segment = audio[s:e]

    # Save phoneme audio
    out_path = f"phoneme_{idx}_{ph}.wav"
    sf.write(out_path, segment, sr)
    output_segments.append(out_path)

    # Visualize waveform
    plt.figure(figsize=(6, 2))
    plt.plot(segment)
    plt.title(f"Phoneme: {ph}  ({t_start:.2f}s – {t_end:.2f}s)")
    plt.xlabel("Samples")
    plt.ylabel("Amplitude")
    plt.show()

# 7. Infer sound source nature (rule-based)
def infer_source(phoneme):
    if phoneme in ["P", "B", "T", "D", "K", "G"]:
        return "Plosive (burst of air, vocal tract closure)"
    elif phoneme in ["S", "Z", "F", "V", "SH"]:
        return "Fricative (turbulent airflow)"
    elif phoneme in ["M", "N", "NG"]:
        return "Nasal (airflow through nasal cavity)"
    elif phoneme in ["AA", "AE", "AH", "EH", "IH", "IY", "OW", "UH", "UW"]:
        return "Vowel (voiced, open vocal tract)"
    else:
        return "Other / approximant sound"

for ph, t1, t2 in phonemes[:5]:
    print(f"Phoneme: {ph}, Source: {infer_source(ph)}")
```

```python
!pip install librosa transformers torch matplotlib

# ================================
# SPEECH PROCESSING LAB
# Word-level Phoneme Demonstration
# ================================

import librosa
import torch
import matplotlib.pyplot as plt
from transformers import Wav2Vec2Processor, Wav2Vec2ForCTC

# ---------- 1. LOAD AUDIO ----------
audio_path = "/content/drive/MyDrive/Speech processing/LJ001-0001.wav"
audio, sr = librosa.load(audio_path, sr=16000, mono=True)

# ---------- 2. LOAD MODEL ----------
processor = Wav2Vec2Processor.from_pretrained("facebook/wav2vec2-base-960h")
model = Wav2Vec2ForCTC.from_pretrained("facebook/wav2vec2-base-960h")
model.eval()

# ---------- 3. SPEECH TO TEXT ----------
inputs = processor(audio, sampling_rate=sr, return_tensors="pt")

with torch.no_grad():
    logits = model(inputs.input_values).logits

predicted_ids = torch.argmax(logits, dim=-1)

# Decode full sentence
decoded_text = processor.batch_decode(predicted_ids)[0]

print("\nDecoded text from audio:")
print(decoded_text)

# ---------- 4. PICK ONE WORD ----------
words = decoded_text.lower().split()
selected_word = words[0]   # take first word

print("\nSelected word for phoneme analysis:")
print(selected_word)

# ---------- 5. TOKEN-LEVEL (PHONEME-LIKE) OUTPUT ----------
tokens = processor.tokenizer.convert_ids_to_tokens(predicted_ids[0])

# ---------- 6. PHONEME SEGMENTATION ----------
time_per_step = len(audio) / logits.shape[1] / sr

phonemes = []
start = 0
prev = tokens[0]

for i, tok in enumerate(tokens):
    if tok != prev:
        if prev != "|":
            phonemes.append((prev, start*time_per_step, i*time_per_step))
        start = i
        prev = tok

if prev != "|":
    phonemes.append((prev, start*time_per_step, len(tokens)*time_per_step))

# ---------- 7. PHONEME CLASSIFICATION ----------
def phoneme_class(ph):
    if ph in ["P","T","K","B","D","G"]:
        return "Plosive"
    elif ph in ["S","SH","F","Z","V","HH"]:
        return "Fricative"
    elif ph in ["M","N","NG"]:
        return "Nasal"
    elif ph in ["L","R","Y","W"]:
        return "Approximant"
    elif ph in ["AA","AE","AH","EH","IH","IY","OW","UH","UW"]:
        return "Vowel"
    else:
        return "Other"

# ---------- 8. TAKE FEW PHONEMES (LIKE HELLO DEMO) ----------
selected_phonemes = phonemes[:4]

print("\nPhonemes used in the selected word:\n")
for ph, _, _ in selected_phonemes:
    print(f"{ph}  →  {phoneme_class(ph)}")

# ---------- 9. PLOT PHONEME WAVEFORMS ----------
print("\nPhoneme Waveforms:\n")

for i, (ph, t1, t2) in enumerate(selected_phonemes):
    s = int(t1 * sr)
    e = int(t2 * sr)
    segment = audio[s:e]

    if len(segment) < 80:
        continue

    plt.figure(figsize=(6,2))
    plt.plot(segment)
    plt.title(f"Phoneme {i+1}: {ph} ({phoneme_class(ph)})")
    plt.xlabel("Samples")
    plt.ylabel("Amplitude")
    plt.tight_layout()
    plt.show()

print("Word-level phoneme analysis completed.")
```
